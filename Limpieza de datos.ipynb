```python
url = "TMDB_movie_dataset_v11.csv"
df = pd.read_csv(url)
df
```

```python
df=df.rename(columns={'title': 'Titulo','vote_average': 'Promedio de voto','vote_count':'Conteo de voto','release_date':'Fecha de lanzamiento',
                      'revenue':'Ingresos', 'runtime':'Duracion','adult':'Adultos',
                      'backdrop_path':'Ruta de fondo','budget':'Presupuesto','homepage':'Pagina principal',
                      'original_language':'Idioma original','original_title':'Titulo original',
                      'overview':'Descripcion','popularity':'Popularidad','poster_path': 'Ruta de poster', 'tagline': 'Eslogan', 'genres':'Genero', 'production_companies':'Productora', 'production_countries':'Pais', 'spoken_languages':'Idiomas', 'keywords':'PL'   })
df
```

```python
df=df.drop(columns=['Ruta de poster', 'Eslogan', 'Ruta de fondo', 'id', 'Pagina principal'])
```

```python
df['Fecha de lanzamiento'] = pd.to_datetime(df['Fecha de lanzamiento'])
df.info()
```

```python
# Extraer el año
df['año'] = df['Fecha de lanzamiento'].dt.year

# Extraer el mes
df['mes'] = df['Fecha de lanzamiento'].dt.month

# Extraer el día
df['dia'] = df['Fecha de lanzamiento'].dt.day

# Extraer el día de la semana
df['díanombre'] = df['Fecha de lanzamiento'].dt.day_name()

df.head()
```

```python
df=df.drop_duplicates()
df
```

```python
lista_col=df.columns
df=df
for i in lista_col:
    df=df[df[i] != 'NaN']
df
```

```python
df.info()
```

```python
df[ "dia" ].fillna(df[ "dia"].mean(), inplace=True)
df[ "mes" ].fillna(df[ "mes"].mean(), inplace=True)
df[ "año" ].fillna(df[ "año"].mean(), inplace=True)	
df[ "Promedio de voto" ].fillna(df[ "Promedio de voto"].mean(), inplace=True)
df[ "Presupuesto" ].fillna(df[ "Presupuesto"].mean(), inplace=True)
df
```

```python
df=df.dropna(subset=['Fecha de lanzamiento'])
df=df.dropna(subset=['díanombre'])
df=df.dropna(subset=['imdb_id'])
df=df.dropna(subset=['Descripcion'])
df=df.dropna(subset=['Genero']) 
df=df.dropna(subset=['Productora']) 
df=df.dropna(subset=['Pais']) 
df=df.dropna(subset=['Idiomas']) 
df=df.dropna(subset=['PL']) 
df
```
